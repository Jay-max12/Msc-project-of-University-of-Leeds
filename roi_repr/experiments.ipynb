# ROI 表征学习五方法对照实验

统一架构：`ROI → ResNet50 → 2048 → Linear(2048→128) → z → Linear(128→C)`

代码包：[`roi_repr/`](../roi_repr/README.md)

修改 `METHOD` / `TASK_MODE` 后 **Restart Kernel → Run All**。

```python
from pathlib import Path
from dataclasses import replace

from roi_repr.config import ReprConfig
from roi_repr.training.trainer import run_experiment, run_all_methods

METHOD = "cls"        # cls | ae | vae | siamese | triplet | all
TASK_MODE = "multiclass"  # multiclass | binary

cfg = ReprConfig(method="cls", task_mode=TASK_MODE)
cfg.__post_init__()
print("output:", cfg.output_dir)

if METHOD == "all":
    results = run_all_methods(cfg)
else:
    cfg = replace(cfg, method=METHOD, output_dir=None)
    cfg.__post_init__()
    results = run_experiment(cfg)

results
```
